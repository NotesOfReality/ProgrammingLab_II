### **1. Trasformare cicli in list comprehension**

Fai tre esempi di cicli producono una lista e poi riscrivi lo stesso codice usando la list comprehension

```python
# 1
meh =  []
for el in "mammamia":
    meh.append(el)
print(meh)
## List comprehension
print([el for el in "mammamia"])

#2
mah = []
for el in range(5):
    mah += [el]
print(mah)
## List comprehension
print([el for el in mah])

#3
from random import randint
eheh = []
for i in range(8):
    eheh[len(eheh):] = [randint(0,10)]
print(eheh)
## List comprehension
print([el for el in eheh])
```

1. **Crea un vettore** contenente tutti i numeri primi compresi tra **0 e 10**  
   _(Puoi scriverli direttamente nell'array)._

2. **Conta quanti numeri ci sono nel vettore** utilizzando la funzione `len()`.  **Ottieni lo stesso numero** accedendo all'attributo `.size` del vettore.

3. Quale pensi sia il tipo di dato (`dtype`) del vettore?  Prova a rispondere senza eseguire il codice e Verifica la tua risposta accedendo all'attributo `.dtype` del vettore

4. Scrivi l'array usando una list comprehension che controlla che i numeri siano primi. 

```python
import math
import numpy as np

# 1
def is_prime(n):
    if n < 2: return False
    for i in range(2, math.isqrt(n) + 1):
        if n % i == 0:
            return False
    return True

primes = np.array([p for p in range(0,11) if is_prime(p)])

# 2
print(len(primes))
print(primes.size)

# 3
# Mi aspetto che .dtype di primes sia int64, perchè il sistema in uso è "a 64 bit", ed i numeri primi sono tutti numeri naturali.
print(np.int_) # np.int_ dovrebbe essere l' alias del tipo di dato predefinito per i numeri naturali
primes.dtype

# 4
# Già fatto in # 1 , heh
```
